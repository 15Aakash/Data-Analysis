```python
import math

control_users = 3048
control_activated = 1552

treatment_users = 3103
treatment_activated = 1574

control_rate = control_activated / control_users
treatment_rate = treatment_activated / treatment_users

absolute_uplift = treatment_rate - control_rate
relative_uplift = absolute_uplift / control_rate

pooled_rate = (
    control_activated + treatment_activated
) / (
    control_users + treatment_users
)

standard_error = math.sqrt(
    pooled_rate
    * (1 - pooled_rate)
    * (
        1 / control_users
        + 1 / treatment_users
    )
)

z_score = absolute_uplift / standard_error

normal_cdf = lambda z: 0.5 * (
    1 + math.erf(z / math.sqrt(2))
)

p_value = 2 * (
    1 - normal_cdf(abs(z_score))
)

print(f"Control activation rate: {control_rate:.2%}")
print(f"Treatment activation rate: {treatment_rate:.2%}")
print(f"Absolute uplift: {absolute_uplift:.2%}")
print(f"Relative uplift: {relative_uplift:.2%}")
print(f"Z-score: {z_score:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
se_difference = math.sqrt(
    (control_rate * (1 - control_rate) / control_users)
    +
    (treatment_rate * (1 - treatment_rate) / treatment_users)
)

ci_lower = absolute_uplift - 1.96 * se_difference
ci_upper = absolute_uplift + 1.96 * se_difference

print(f"95% CI lower bound: {ci_lower:.2%}")
print(f"95% CI upper bound: {ci_upper:.2%}")
```
