```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

RAW_DATA_DIR
```

```python
# -------------------------------------------------
# Raw data inventory
# -------------------------------------------------

expected_files = [
    "users.csv",
    "organizations.csv",
    "plans.csv",
    "subscriptions.csv",
    "payments.csv",
    "sessions.csv",
    "features.csv",
    "product_events.csv",
    "onboarding_events.csv",
    "marketing_attribution.csv",
    "support_tickets.csv",
    "experiments.csv",
    "experiment_assignments.csv"
]

inventory = []

for file_name in expected_files:

    file_path = RAW_DATA_DIR / file_name

    exists = file_path.exists()

    row_count = None

    if exists:
        with open(
            file_path,
            "r",
            encoding="utf-8"
        ) as f:
            row_count = sum(1 for _ in f) - 1

    inventory.append({
        "file_name": file_name,
        "exists": exists,
        "row_count": row_count
    })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

### Summary check

```python
print(
    "Expected files:",
    len(expected_files)
)

print(
    "Files found:",
    inventory_df[
        "exists"
    ].sum()
)

print(
    "Missing files:",
    (
        ~inventory_df[
            "exists"
        ]
    ).sum()
)

print()

print(
    inventory_df[
        [
            "file_name",
            "row_count"
        ]
    ]
)
```

### Check total raw volume

```python
total_rows = (
    inventory_df[
        "row_count"
    ]
    .fillna(0)
    .sum()
)

print(
    "Total raw rows across all files:",
    f"{int(total_rows):,}"
)
```

### Save the inventory

```python
inventory_file = (
    PROJECT_ROOT
    / "docs"
    / "raw_data_inventory.csv"
)

inventory_df.to_csv(
    inventory_file,
    index=False
)

print(
    f"Saved: {inventory_file}"
)
```
