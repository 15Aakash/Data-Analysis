## Add controlled data-quality problems

### Create the imperfect raw copy

```python
# -------------------------------------------------
# Create raw users dataset with controlled
# data-quality issues
# -------------------------------------------------

users_raw_df = users_df.copy()

# Separate random generator for reproducible DQ issues
dq_rng = np.random.default_rng(SEED + 100)

# Shuffle row indexes so each issue can be placed
# on a different set of users
available_indexes = np.arange(len(users_raw_df))
dq_rng.shuffle(available_indexes)

cursor = 0

def get_indexes(n):
    global cursor
    selected = available_indexes[cursor:cursor + n]
    cursor += n
    return selected
```

### Introduce the issues

```python
# -------------------------------------------------
# Inject realistic raw data-quality issues
# -------------------------------------------------

# 1. Invalid email formats
invalid_email_idx = get_indexes(120)
users_raw_df.loc[invalid_email_idx, "email"] = (
    "invalid_email_" +
    users_raw_df.loc[invalid_email_idx, "user_id"]
)

# 2. Missing countries
missing_country_idx = get_indexes(100)
users_raw_df.loc[missing_country_idx, "country"] = None

# 3. Invalid signup platforms
invalid_platform_idx = get_indexes(80)
users_raw_df.loc[
    invalid_platform_idx,
    "signup_platform"
] = "DesktopApp"

# 4. Invalid account statuses
invalid_status_idx = get_indexes(60)
users_raw_df.loc[
    invalid_status_idx,
    "account_status"
] = "Unknown"

# 5. Future registration timestamps
future_date_idx = get_indexes(50)
users_raw_df.loc[
    future_date_idx,
    "registration_timestamp"
] = pd.Timestamp("2027-01-15")

# 6. Broken organization foreign keys
broken_org_idx = get_indexes(50)
users_raw_df.loc[
    broken_org_idx,
    "organization_id"
] = "ORG99999"

print("Controlled data-quality issues added.")
```

### Add duplicate records

```python
# -------------------------------------------------
# Add duplicate user records
# -------------------------------------------------

duplicate_count = 75

duplicate_rows = users_raw_df.sample(
    n=duplicate_count,
    random_state=SEED
).copy()

users_raw_df = pd.concat(
    [users_raw_df, duplicate_rows],
    ignore_index=True
)

print("Original clean rows:", len(users_df))
print("Raw rows after duplicates:", len(users_raw_df))
```

### Generate subscriptions.csv

```python
# -------------------------------------------------
# Generate user-level subscriptions
# -------------------------------------------------

CURRENT_DATE = pd.Timestamp("2026-08-31")

N_USER_SUBSCRIPTIONS = 66_000

subscription_rng = np.random.default_rng(SEED + 200)

# Select users who will have a subscription
selected_user_indexes = subscription_rng.choice(
    users_df.index,
    size=N_USER_SUBSCRIPTIONS,
    replace=False
)

selected_users = users_df.loc[
    selected_user_indexes,
    ["user_id", "registration_timestamp"]
].reset_index(drop=True)

user_subscription_rows = []

for i, row in selected_users.iterrows():

    user_id = row["user_id"]
    registration_date = pd.Timestamp(
        row["registration_timestamp"]
    )

    # Subscription cannot start before registration
    max_days = max(
        (CURRENT_DATE - registration_date).days,
        0
    )

    start_offset = (
        subscription_rng.integers(0, max_days + 1)
        if max_days > 0
        else 0
    )

    start_date = (
        registration_date
        + pd.Timedelta(days=int(start_offset))
    ).normalize()

    # Individual-user plan mix
    plan_id = subscription_rng.choice(
        ["P001", "P002", "P003"],
        p=[0.48, 0.47, 0.05]
    )

    billing_cycle = subscription_rng.choice(
        ["Monthly", "Annual"],
        p=[0.72, 0.28]
    )

    status = subscription_rng.choice(
        ["Active", "Cancelled", "Expired", "Trial"],
        p=[0.64, 0.18, 0.10, 0.08]
    )

    end_date = pd.NaT
    cancellation_reason = None

    if status in ["Cancelled", "Expired"]:

        available_days = max(
            (CURRENT_DATE - start_date).days,
            1
        )

        end_offset = subscription_rng.integers(
            1,
            available_days + 1
        )

        end_date = (
            start_date
            + pd.Timedelta(days=int(end_offset))
        )

        cancellation_reason = (
            subscription_rng.choice(
                [
                    "Too Expensive",
                    "Low Usage",
                    "Missing Features",
                    "Switched Competitor",
                    "Temporary Need",
                    "Other"
                ]
            )
            if status == "Cancelled"
            else None
        )

    auto_renew_flag = bool(
        subscription_rng.random()
        <
        (
            0.88 if status == "Active"
            else 0.65 if status == "Trial"
            else 0.05
        )
    )

    user_subscription_rows.append({
        "user_id": user_id,
        "organization_id": None,
        "plan_id": plan_id,
        "subscription_start_date": start_date.date(),
        "subscription_end_date": (
            end_date.date()
            if pd.notna(end_date)
            else None
        ),
        "billing_cycle": billing_cycle,
        "subscription_status": status,
        "auto_renew_flag": auto_renew_flag,
        "cancellation_reason": cancellation_reason
    })

user_subscriptions_df = pd.DataFrame(
    user_subscription_rows
)

user_subscriptions_df.head()
```

### Generate organization subscriptions

```python
# -------------------------------------------------
# Generate organization-level subscriptions
# -------------------------------------------------

N_ORG_SUBSCRIPTIONS = 6_000

selected_org_indexes = subscription_rng.choice(
    organizations_df.index,
    size=N_ORG_SUBSCRIPTIONS,
    replace=False
)

selected_orgs = organizations_df.loc[
    selected_org_indexes,
    ["organization_id", "created_timestamp"]
].reset_index(drop=True)

org_subscription_rows = []

for i, row in selected_orgs.iterrows():

    organization_id = row["organization_id"]

    created_date = pd.Timestamp(
        row["created_timestamp"]
    )

    max_days = max(
        (CURRENT_DATE - created_date).days,
        0
    )

    start_offset = (
        subscription_rng.integers(0, max_days + 1)
        if max_days > 0
        else 0
    )

    start_date = (
        created_date
        + pd.Timedelta(days=int(start_offset))
    ).normalize()

    # Organizations are primarily Business customers
    plan_id = subscription_rng.choice(
        ["P002", "P003"],
        p=[0.20, 0.80]
    )

    billing_cycle = subscription_rng.choice(
        ["Monthly", "Annual"],
        p=[0.40, 0.60]
    )

    status = subscription_rng.choice(
        ["Active", "Cancelled", "Expired", "Trial"],
        p=[0.72, 0.14, 0.08, 0.06]
    )

    end_date = pd.NaT
    cancellation_reason = None

    if status in ["Cancelled", "Expired"]:

        available_days = max(
            (CURRENT_DATE - start_date).days,
            1
        )

        end_offset = subscription_rng.integers(
            1,
            available_days + 1
        )

        end_date = (
            start_date
            + pd.Timedelta(days=int(end_offset))
        )

        cancellation_reason = (
            subscription_rng.choice(
                [
                    "Too Expensive",
                    "Low Usage",
                    "Missing Features",
                    "Switched Competitor",
                    "Contract Ended",
                    "Other"
                ]
            )
            if status == "Cancelled"
            else None
        )

    auto_renew_flag = bool(
        subscription_rng.random()
        <
        (
            0.92 if status == "Active"
            else 0.70 if status == "Trial"
            else 0.05
        )
    )

    org_subscription_rows.append({
        "user_id": None,
        "organization_id": organization_id,
        "plan_id": plan_id,
        "subscription_start_date": start_date.date(),
        "subscription_end_date": (
            end_date.date()
            if pd.notna(end_date)
            else None
        ),
        "billing_cycle": billing_cycle,
        "subscription_status": status,
        "auto_renew_flag": auto_renew_flag,
        "cancellation_reason": cancellation_reason
    })

org_subscriptions_df = pd.DataFrame(
    org_subscription_rows
)

org_subscriptions_df.head()
```

### Combine them and create the subscription PK

```python
# -------------------------------------------------
# Combine user and organization subscriptions
# -------------------------------------------------

subscriptions_df = pd.concat(
    [
        user_subscriptions_df,
        org_subscriptions_df
    ],
    ignore_index=True
)

subscriptions_df.insert(
    0,
    "subscription_id",
    [
        f"SUB{str(i).zfill(6)}"
        for i in range(
            1,
            len(subscriptions_df) + 1
        )
    ]
)

subscriptions_df.head()
```

```python
# -------------------------------------------------
# Generate payment transactions
# -------------------------------------------------

payment_rng = np.random.default_rng(SEED + 300)

payment_rows = []

paid_subscriptions = subscriptions_df[
    (subscriptions_df["plan_id"] != "P001")
    &
    (subscriptions_df["subscription_status"] != "Trial")
].copy()

for _, row in paid_subscriptions.iterrows():

    subscription_id = row["subscription_id"]
    plan_id = row["plan_id"]
    billing_cycle = row["billing_cycle"]

    start_date = pd.Timestamp(
        row["subscription_start_date"]
    )

    if pd.notna(row["subscription_end_date"]):
        effective_end_date = min(
            pd.Timestamp(row["subscription_end_date"]),
            CURRENT_DATE
        )
    else:
        effective_end_date = CURRENT_DATE

    # Skip impossible or zero-length periods
    if effective_end_date < start_date:
        continue

    if billing_cycle == "Monthly":

        amount = float(
            plan_price_lookup[
                plan_id
            ]["monthly_price"]
        )

        payment_dates = pd.date_range(
            start=start_date,
            end=effective_end_date,
            freq=pd.DateOffset(months=1)
        )

    else:

        amount = float(
            plan_price_lookup[
                plan_id
            ]["annual_price"]
        )

        payment_dates = pd.date_range(
            start=start_date,
            end=effective_end_date,
            freq=pd.DateOffset(years=1)
        )

    for payment_date in payment_dates:

        payment_status = payment_rng.choice(
            [
                "Successful",
                "Failed",
                "Refunded"
            ],
            p=[
                0.94,
                0.04,
                0.02
            ]
        )

        payment_method = payment_rng.choice(
            [
                "Card",
                "PayPal",
                "Bank Transfer"
            ],
            p=[
                0.72,
                0.18,
                0.10
            ]
        )

        payment_rows.append({
            "subscription_id": subscription_id,
            "payment_timestamp": (
                payment_date
                + pd.Timedelta(
                    hours=int(
                        payment_rng.integers(0, 24)
                    ),
                    minutes=int(
                        payment_rng.integers(0, 60)
                    )
                )
            ),
            "amount": amount,
            "currency": "USD",
            "payment_status": payment_status,
            "payment_method_type": payment_method
        })

payments_df = pd.DataFrame(payment_rows)

payments_df.head()
```

```python
# -------------------------------------------------
# Assign synthetic engagement segments
# -------------------------------------------------

user_behavior_df["engagement_segment"] = None


# Active paid customers
mask = (
    user_behavior_df["plan_id"].isin(["P002", "P003"])
    &
    (user_behavior_df["subscription_status"] == "Active")
)

user_behavior_df.loc[
    mask,
    "engagement_segment"
] = session_rng.choice(
    ["High", "Medium", "Low", "Dormant"],
    size=mask.sum(),
    p=[0.35, 0.45, 0.15, 0.05]
)


# Cancelled / expired customers
mask = user_behavior_df[
    "subscription_status"
].isin(["Cancelled", "Expired"])

user_behavior_df.loc[
    mask,
    "engagement_segment"
] = session_rng.choice(
    ["High", "Medium", "Low", "Dormant"],
    size=mask.sum(),
    p=[0.08, 0.22, 0.40, 0.30]
)


# Trial customers
mask = (
    user_behavior_df["subscription_status"]
    == "Trial"
)

user_behavior_df.loc[
    mask,
    "engagement_segment"
] = session_rng.choice(
    ["High", "Medium", "Low", "Dormant"],
    size=mask.sum(),
    p=[0.20, 0.40, 0.30, 0.10]
)


# Free-plan customers
mask = (
    (user_behavior_df["plan_id"] == "P001")
    &
    user_behavior_df[
        "engagement_segment"
    ].isna()
)

user_behavior_df.loc[
    mask,
    "engagement_segment"
] = session_rng.choice(
    ["High", "Medium", "Low", "Dormant"],
    size=mask.sum(),
    p=[0.12, 0.38, 0.35, 0.15]
)


# Users without any subscription
mask = user_behavior_df[
    "engagement_segment"
].isna()

user_behavior_df.loc[
    mask,
    "engagement_segment"
] = session_rng.choice(
    ["High", "Medium", "Low", "Dormant"],
    size=mask.sum(),
    p=[0.08, 0.30, 0.40, 0.22]
)
```

### Determine how many sessions each user gets

```python
# -------------------------------------------------
# Generate number of sessions per user
# -------------------------------------------------

session_lambda_map = {
    "High": 20.0,
    "Medium": 10.0,
    "Low": 4.0,
    "Dormant": 0.6
}

user_behavior_df[
    "base_session_lambda"
] = user_behavior_df[
    "engagement_segment"
].map(session_lambda_map)


# Older accounts have had more time to generate sessions
registration_dates = pd.to_datetime(
    user_behavior_df[
        "registration_timestamp"
    ]
)

account_age_days = (
    CURRENT_DATE
    - registration_dates.dt.normalize()
).dt.days.clip(lower=0)

max_account_age = max(
    account_age_days.max(),
    1
)

age_factor = (
    0.40
    +
    0.60
    * (
        account_age_days
        / max_account_age
    )
)


# Reduce expected activity for suspended/deleted users
status_multiplier = (
    user_behavior_df[
        "account_status"
    ]
    .map({
        "Active": 1.00,
        "Suspended": 0.40,
        "Deleted": 0.15
    })
    .fillna(1.00)
)


effective_lambda = (
    user_behavior_df[
        "base_session_lambda"
    ]
    * age_factor
    * status_multiplier
)


user_behavior_df[
    "session_count"
] = session_rng.poisson(
    effective_lambda
)

print(
    "Expected total sessions:",
    f"{user_behavior_df['session_count'].sum():,}"
)
```

### Determine each user's effective plan

```python
# -------------------------------------------------
# Determine effective user plan
# -------------------------------------------------

direct_user_plans = (
    subscriptions_df[
        subscriptions_df["user_id"].notna()
    ]
    .drop_duplicates(
        subset="user_id",
        keep="last"
    )[
        ["user_id", "plan_id"]
    ]
    .rename(
        columns={
            "plan_id": "direct_plan_id"
        }
    )
)

organization_plans = (
    subscriptions_df[
        subscriptions_df["organization_id"].notna()
    ]
    .drop_duplicates(
        subset="organization_id",
        keep="last"
    )[
        ["organization_id", "plan_id"]
    ]
    .rename(
        columns={
            "plan_id": "organization_plan_id"
        }
    )
)

user_plan_df = (
    users_df[
        ["user_id", "organization_id"]
    ]
    .merge(
        direct_user_plans,
        on="user_id",
        how="left"
    )
    .merge(
        organization_plans,
        on="organization_id",
        how="left"
    )
)

user_plan_df["effective_plan_id"] = (
    user_plan_df["direct_plan_id"]
    .fillna(
        user_plan_df["organization_plan_id"]
    )
    .fillna("P001")
)

user_plan_df[
    ["user_id", "effective_plan_id"]
].head()
```

```python
sessions_events_base = sessions_df.merge(
    user_plan_df[
        [
            "user_id",
            "effective_plan_id"
        ]
    ],
    on="user_id",
    how="left"
)

sessions_events_base[
    "effective_plan_id"
] = sessions_events_base[
    "effective_plan_id"
].fillna("P001")

sessions_events_base.head()
```

### Calculate events per session

```python
# -------------------------------------------------
# Determine number of product events per session
# -------------------------------------------------

event_rng = np.random.default_rng(SEED + 700)

session_duration_minutes = (
    (
        sessions_events_base[
            "session_end_timestamp"
        ]
        -
        sessions_events_base[
            "session_start_timestamp"
        ]
    )
    .dt.total_seconds()
    .div(60)
    .clip(lower=0)
)

event_lambda = (
    2.6
    +
    np.minimum(
        session_duration_minutes / 60,
        1.5
    )
)

# Abandoned sessions have fewer actions
abandoned_mask = (
    sessions_events_base[
        "session_status"
    ]
    == "Abandoned"
)

event_lambda.loc[
    abandoned_mask
] = 1.4

sessions_events_base[
    "event_count"
] = np.maximum(
    1,
    event_rng.poisson(
        event_lambda.to_numpy()
    )
)

TOTAL_EVENTS = int(
    sessions_events_base[
        "event_count"
    ].sum()
)

print(
    "Estimated product events:",
    f"{TOTAL_EVENTS:,}"
)
```

### Define feature → event mapping

```python
feature_event_map = {
    "F001": "project_created",
    "F002": "task_updated",
    "F003": "file_uploaded",
    "F004": "collaboration_action",
    "F005": "calendar_synced",
    "F006": "automation_run",
    "F007": "dashboard_viewed",
    "F008": "analytics_report_viewed",
    "F009": "permission_changed",
    "F010": "audit_log_viewed",
    "F011": "api_request",
    "F012": "ai_assistant_used"
}

feature_weights = {
    "F001": 0.16,
    "F002": 0.20,
    "F003": 0.15,
    "F004": 0.10,
    "F005": 0.08,
    "F006": 0.08,
    "F007": 0.07,
    "F008": 0.04,
    "F009": 0.03,
    "F010": 0.03,
    "F011": 0.03,
    "F012": 0.03
}
```

### Prepare feature eligibility

```python
plan_rank = {
    "P001": 1,
    "P002": 2,
    "P003": 3
}

minimum_plan_rank = {
    "Free": 1,
    "Pro": 2,
    "Business": 3
}

feature_meta = features_df.copy()

feature_meta[
    "minimum_plan_rank"
] = feature_meta[
    "minimum_plan"
].map(minimum_plan_rank)

feature_meta[
    "launch_timestamp"
] = pd.to_datetime(
    feature_meta["launch_date"]
)

feature_meta[
    [
        "feature_id",
        "minimum_plan",
        "launch_timestamp"
    ]
]
```

### Generate the large CSV in chunks

```python
# -------------------------------------------------
# Generate product events in chunks
# -------------------------------------------------

PRODUCT_EVENTS_FILE = (
    RAW_DATA_DIR / "product_events.csv"
)

CHUNK_SIZE = 50_000

event_counter = 0
first_chunk = True

generic_event_names = [
    "login",
    "navigation_view",
    "search_used"
]

for chunk_start in range(
    0,
    len(sessions_events_base),
    CHUNK_SIZE
):

    chunk = sessions_events_base.iloc[
        chunk_start:
        chunk_start + CHUNK_SIZE
    ].reset_index(drop=True)

    counts = (
        chunk["event_count"]
        .to_numpy()
    )

    repeated_indexes = np.repeat(
        np.arange(len(chunk)),
        counts
    )

    n_events = len(repeated_indexes)

    session_ids = (
        chunk["session_id"]
        .to_numpy()[repeated_indexes]
    )

    user_ids = (
        chunk["user_id"]
        .to_numpy()[repeated_indexes]
    )

    platforms = (
        chunk["platform"]
        .to_numpy()[repeated_indexes]
    )

    plans = (
        chunk["effective_plan_id"]
        .to_numpy()[repeated_indexes]
    )

    starts = pd.to_datetime(
        chunk[
            "session_start_timestamp"
        ]
        .to_numpy()[repeated_indexes]
    )

    ends = pd.to_datetime(
        chunk[
            "session_end_timestamp"
        ]
        .to_numpy()[repeated_indexes]
    )

    duration_seconds = (
        (ends - starts)
        .total_seconds()
    )

    offsets = (
        event_rng.random(n_events)
        * duration_seconds
    )

    event_timestamps = (
        starts
        +
        pd.to_timedelta(
            offsets,
            unit="s"
        )
    )

    # ---------------------------------------------
    # Generic vs feature-specific events
    # ---------------------------------------------

    generic_mask = (
        event_rng.random(n_events)
        < 0.20
    )

    feature_ids = np.full(
        n_events,
        None,
        dtype=object
    )

    event_names = np.empty(
        n_events,
        dtype=object
    )

    event_names[
        generic_mask
    ] = event_rng.choice(
        generic_event_names,
        size=generic_mask.sum(),
        p=[0.45, 0.35, 0.20]
    )

    # ---------------------------------------------
    # Generate feature events by plan + launch date
    # ---------------------------------------------

    feature_event_indexes = np.where(
        ~generic_mask
    )[0]

    for plan_id in [
        "P001",
        "P002",
        "P003"
    ]:

        plan_indexes = feature_event_indexes[
            plans[
                feature_event_indexes
            ]
            == plan_id
        ]

        if len(plan_indexes) == 0:
            continue

        rank = plan_rank[plan_id]

        eligible_features = feature_meta[
            feature_meta[
                "minimum_plan_rank"
            ] <= rank
        ]

        for event_index in plan_indexes:

            timestamp = (
                event_timestamps[
                    event_index
                ]
            )

            available_features = (
                eligible_features[
                    eligible_features[
                        "launch_timestamp"
                    ]
                    <= timestamp
                ]
            )

            # Very early sessions may occur before
            # any feature was officially launched
            if available_features.empty:

                feature_ids[
                    event_index
                ] = None

                event_names[
                    event_index
                ] = event_rng.choice(
                    generic_event_names
                )

                continue

            candidate_ids = (
                available_features[
                    "feature_id"
                ]
                .tolist()
            )

            candidate_weights = np.array(
                [
                    feature_weights[
                        feature_id
                    ]
                    for feature_id
                    in candidate_ids
                ]
            )

            candidate_weights = (
                candidate_weights
                /
                candidate_weights.sum()
            )

            selected_feature = (
                event_rng.choice(
                    candidate_ids,
                    p=candidate_weights
                )
            )

            feature_ids[
                event_index
            ] = selected_feature

            event_names[
                event_index
            ] = feature_event_map[
                selected_feature
            ]

    # ---------------------------------------------
    # Build chunk DataFrame
    # ---------------------------------------------

    event_ids = [
        f"E{str(i).zfill(8)}"
        for i in range(
            event_counter + 1,
            event_counter
            + n_events
            + 1
        )
    ]

    events_chunk_df = pd.DataFrame({
        "event_id": event_ids,
        "user_id": user_ids,
        "session_id": session_ids,
        "feature_id": feature_ids,
        "event_name": event_names,
        "event_timestamp": event_timestamps,
        "platform": platforms,
        "event_value": None
    })

    events_chunk_df.to_csv(
        PRODUCT_EVENTS_FILE,
        mode=(
            "w"
            if first_chunk
            else "a"
        ),
        header=first_chunk,
        index=False
    )

    event_counter += n_events
    first_chunk = False

    print(
        f"Generated {event_counter:,} events..."
    )

print()
print(
    "Final product event rows:",
    f"{event_counter:,}"
)
```

```python
# -------------------------------------------------
# Assign onboarding completion depth
# -------------------------------------------------

onboarding_rng = np.random.default_rng(SEED + 800)

onboarding_users = users_df[
    [
        "user_id",
        "registration_timestamp",
        "signup_platform"
    ]
].copy()

completion_depths = onboarding_rng.choice(
    [1, 2, 3, 4, 5, 6],
    size=len(onboarding_users),
    p=[
        0.04,
        0.06,
        0.09,
        0.12,
        0.18,
        0.51
    ]
)

onboarding_users[
    "completion_depth"
] = completion_depths

onboarding_users[
    "completion_depth"
].value_counts(
    normalize=True
).sort_index().mul(100).round(2)
```

```python
# -------------------------------------------------
# Generate onboarding events
# -------------------------------------------------

onboarding_rows = []

for _, user in onboarding_users.iterrows():

    user_id = user["user_id"]

    registration_timestamp = pd.Timestamp(
        user["registration_timestamp"]
    )

    completion_depth = int(
        user["completion_depth"]
    )

    current_timestamp = (
        registration_timestamp
        +
        pd.Timedelta(
            minutes=int(
                onboarding_rng.integers(
                    1,
                    20
                )
            )
        )
    )

    for step_order, step_name in onboarding_steps:

        if step_order > completion_depth:
            break

        time_spent_seconds = int(
            onboarding_rng.integers(
                30,
                900
            )
        )

        # Last reached step may be abandoned
        if (
            step_order == completion_depth
            and completion_depth < 6
        ):
            completion_status = onboarding_rng.choice(
                [
                    "Completed",
                    "Abandoned"
                ],
                p=[
                    0.20,
                    0.80
                ]
            )
        else:
            completion_status = "Completed"

        onboarding_rows.append({
            "user_id": user_id,
            "onboarding_step": step_name,
            "step_order": step_order,
            "event_timestamp": current_timestamp,
            "completion_status": completion_status,
            "time_spent_seconds": time_spent_seconds
        })

        current_timestamp += pd.Timedelta(
            seconds=time_spent_seconds
        )

        current_timestamp += pd.Timedelta(
            minutes=int(
                onboarding_rng.integers(
                    1,
                    240
                )
            )
        )

onboarding_events_df = pd.DataFrame(
    onboarding_rows
)

print(
    "Generated onboarding events:",
    f"{len(onboarding_events_df):,}"
)

onboarding_events_df.head()
```

## Generate support_tickets.csv

### Prepare user support profile

```python
# -------------------------------------------------
# Prepare user support profile
# -------------------------------------------------

support_rng = np.random.default_rng(SEED + 1000)

support_user_df = (
    user_behavior_df[
        [
            "user_id",
            "engagement_segment",
            "plan_id",
            "subscription_status"
        ]
    ]
    .copy()
)

support_user_df["plan_id"] = (
    support_user_df["plan_id"]
    .fillna("P001")
)

support_user_df.head()
```

### Decide how many support tickets each user creates

```python
# -------------------------------------------------
# Generate ticket count per user
# -------------------------------------------------

support_lambda_by_plan = {
    "P001": 0.18,
    "P002": 0.38,
    "P003": 0.65
}

support_user_df["ticket_lambda"] = (
    support_user_df["plan_id"]
    .map(support_lambda_by_plan)
    .fillna(0.18)
)

# Cancelled / expired users tend to have slightly
# more support interactions in this simulation
churn_related_mask = (
    support_user_df[
        "subscription_status"
    ].isin(
        ["Cancelled", "Expired"]
    )
)

support_user_df.loc[
    churn_related_mask,
    "ticket_lambda"
] *= 1.35

support_user_df["ticket_count"] = (
    support_rng.poisson(
        support_user_df[
            "ticket_lambda"
        ].to_numpy()
    )
)

print(
    "Expected support tickets:",
    f"{support_user_df['ticket_count'].sum():,}"
)
```

### Expand users into ticket rows

```python
# -------------------------------------------------
# Expand users into support ticket records
# -------------------------------------------------

ticket_counts = (
    support_user_df[
        "ticket_count"
    ].to_numpy()
)

support_user_indexes = np.repeat(
    np.arange(
        len(support_user_df)
    ),
    ticket_counts
)

N_TICKETS = len(
    support_user_indexes
)

print(
    "Generating tickets:",
    f"{N_TICKETS:,}"
)
```

### Define ticket categories

```python
# -------------------------------------------------
# Define support categories and priorities
# -------------------------------------------------

ticket_categories = [
    "Billing",
    "Product Issue",
    "Account",
    "Feature Request",
    "Integration",
    "Performance",
    "Security"
]

category_probabilities = [
    0.18,
    0.27,
    0.14,
    0.15,
    0.11,
    0.10,
    0.05
]

ticket_priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]
```

### Generate ticket creation dates

```python
# -------------------------------------------------
# Generate ticket creation timestamps
# -------------------------------------------------

ticket_user_ids = (
    support_user_df[
        "user_id"
    ]
    .to_numpy()[
        support_user_indexes
    ]
)

ticket_user_lookup = (
    users_df
    .set_index("user_id")[
        "registration_timestamp"
    ]
)

ticket_registration_dates = (
    pd.to_datetime(
        pd.Series(
            ticket_user_ids
        ).map(
            ticket_user_lookup
        )
    )
)

days_available = (
    CURRENT_DATE
    -
    ticket_registration_dates.dt.normalize()
).dt.days.clip(
    lower=0
)

ticket_day_offsets = (
    support_rng.random(
        N_TICKETS
    )
    *
    (days_available + 1)
).astype(int)

ticket_seconds = support_rng.integers(
    0,
    24 * 60 * 60,
    size=N_TICKETS
)

created_timestamps = (
    ticket_registration_dates.dt.normalize()
    +
    pd.to_timedelta(
        ticket_day_offsets,
        unit="D"
    )
    +
    pd.to_timedelta(
        ticket_seconds,
        unit="s"
    )
)
```

### Generate category and priority

```python
# -------------------------------------------------
# Generate ticket category and priority
# -------------------------------------------------

generated_categories = support_rng.choice(
    ticket_categories,
    size=N_TICKETS,
    p=category_probabilities
)

generated_priorities = support_rng.choice(
    ticket_priorities,
    size=N_TICKETS,
    p=[
        0.32,
        0.46,
        0.18,
        0.04
    ]
)
```

### Generate resolution times

```python
# -------------------------------------------------
# Generate resolution time
# -------------------------------------------------

resolution_ranges = {
    "Low": (4, 72),
    "Medium": (2, 48),
    "High": (1, 24),
    "Critical": (0.5, 8)
}

resolution_hours = np.zeros(
    N_TICKETS
)

for priority in ticket_priorities:

    priority_mask = (
        generated_priorities
        == priority
    )

    low, high = (
        resolution_ranges[
            priority
        ]
    )

    resolution_hours[
        priority_mask
    ] = support_rng.uniform(
        low,
        high,
        size=priority_mask.sum()
    )

resolution_hours = np.round(
    resolution_hours,
    2
)
```

### Generate ticket status

```python
# -------------------------------------------------
# Generate ticket status
# -------------------------------------------------

ticket_statuses = support_rng.choice(
    [
        "Open",
        "In Progress",
        "Resolved",
        "Closed"
    ],
    size=N_TICKETS,
    p=[
        0.05,
        0.07,
        0.38,
        0.50
    ]
)

resolved_mask = np.isin(
    ticket_statuses,
    [
        "Resolved",
        "Closed"
    ]
)
```

### Calculate resolved timestamps

```python
# -------------------------------------------------
# Generate resolved timestamps
# -------------------------------------------------

resolved_timestamps = pd.Series(
    pd.NaT,
    index=np.arange(N_TICKETS),
    dtype="datetime64[ns]"
)

resolved_timestamps.loc[
    resolved_mask
] = (
    created_timestamps.loc[
        resolved_mask
    ]
    +
    pd.to_timedelta(
        resolution_hours[
            resolved_mask
        ],
        unit="h"
    )
)
```

### Customer satisfaction

```python
# -------------------------------------------------
# Generate satisfaction scores
# -------------------------------------------------

satisfaction_scores = np.full(
    N_TICKETS,
    np.nan
)

for i in np.where(
    resolved_mask
)[0]:

    hours = resolution_hours[i]

    if hours <= 8:
        probabilities = [
            0.02,
            0.05,
            0.13,
            0.35,
            0.45
        ]

    elif hours <= 24:
        probabilities = [
            0.05,
            0.10,
            0.25,
            0.38,
            0.22
        ]

    else:
        probabilities = [
            0.14,
            0.22,
            0.32,
            0.22,
            0.10
        ]

    satisfaction_scores[i] = (
        support_rng.choice(
            [1, 2, 3, 4, 5],
            p=probabilities
        )
    )
```

### Build the table

```python
# -------------------------------------------------
# Build support tickets table
# -------------------------------------------------

support_tickets_df = pd.DataFrame({
    "ticket_id": [
        f"TKT{str(i).zfill(6)}"
        for i in range(
            1,
            N_TICKETS + 1
        )
    ],
    "user_id": ticket_user_ids,
    "ticket_category":
        generated_categories,
    "priority":
        generated_priorities,
    "created_timestamp":
        created_timestamps,
    "resolved_timestamp":
        resolved_timestamps,
    "ticket_status":
        ticket_statuses,
    "satisfaction_score":
        satisfaction_scores,
    "resolution_time_hours":
        np.where(
            resolved_mask,
            resolution_hours,
            np.nan
        )
})

support_tickets_df.head()
```

## Generate experiment_assignments.csv

### Get eligible users for EXP001

```python
# -------------------------------------------------
# Prepare eligible users for EXP001
# -------------------------------------------------

experiment_rng = np.random.default_rng(SEED + 1100)

exp001_start = pd.Timestamp("2026-04-01")
exp001_end = pd.Timestamp("2026-05-31 23:59:59")

eligible_exp001_users = users_df[
    (
        pd.to_datetime(
            users_df["registration_timestamp"]
        ) >= exp001_start
    )
    &
    (
        pd.to_datetime(
            users_df["registration_timestamp"]
        ) <= exp001_end
    )
].copy()

print(
    "Eligible EXP001 users:",
    f"{len(eligible_exp001_users):,}"
)
```

### Assign Control or Treatment

```python
# -------------------------------------------------
# Assign EXP001 experiment groups
# -------------------------------------------------

eligible_exp001_users[
    "experiment_group"
] = experiment_rng.choice(
    ["Control", "Treatment"],
    size=len(eligible_exp001_users),
    p=[0.50, 0.50]
)

eligible_exp001_users[
    "experiment_group"
].value_counts()
```

### Generate assignment/exposure timestamps

```python
# -------------------------------------------------
# Generate assignment and exposure timestamps
# -------------------------------------------------

registration_ts = pd.to_datetime(
    eligible_exp001_users[
        "registration_timestamp"
    ]
)

assignment_delay_minutes = (
    experiment_rng.integers(
        0,
        60,
        size=len(
            eligible_exp001_users
        )
    )
)

eligible_exp001_users[
    "assigned_timestamp"
] = (
    registration_ts
    +
    pd.to_timedelta(
        assignment_delay_minutes,
        unit="m"
    )
)

exposure_delay_minutes = (
    experiment_rng.integers(
        1,
        180,
        size=len(
            eligible_exp001_users
        )
    )
)

eligible_exp001_users[
    "exposure_timestamp"
] = (
    eligible_exp001_users[
        "assigned_timestamp"
    ]
    +
    pd.to_timedelta(
        exposure_delay_minutes,
        unit="m"
    )
)
```

### Build experiment_assignments_df

```python
# -------------------------------------------------
# Build experiment assignments table
# -------------------------------------------------

experiment_assignments_df = pd.DataFrame({
    "experiment_id": "EXP001",
    "user_id":
        eligible_exp001_users[
            "user_id"
        ].values,
    "experiment_group":
        eligible_exp001_users[
            "experiment_group"
        ].values,
    "assigned_timestamp":
        eligible_exp001_users[
            "assigned_timestamp"
        ].values,
    "exposure_timestamp":
        eligible_exp001_users[
            "exposure_timestamp"
        ].values
})

experiment_assignments_df.insert(
    0,
    "assignment_id",
    [
        f"ASN{str(i).zfill(7)}"
        for i in range(
            1,
            len(
                experiment_assignments_df
            ) + 1
        )
    ]
)

experiment_assignments_df.head()
```

### Validate the experiment assignment rules

```python
# -------------------------------------------------
# Validate experiment assignments
# -------------------------------------------------

print(
    "Total assignments:",
    f"{len(experiment_assignments_df):,}"
)

print(
    "Unique assignment IDs:",
    experiment_assignments_df[
        "assignment_id"
    ].nunique()
)

print(
    "Invalid user references:",
    (
        ~experiment_assignments_df[
            "user_id"
        ].isin(
            users_df["user_id"]
        )
    ).sum()
)

print(
    "Invalid experiment references:",
    (
        ~experiment_assignments_df[
            "experiment_id"
        ].isin(
            experiments_df[
                "experiment_id"
            ]
        )
    ).sum()
)

print(
    "Duplicate user-experiment pairs:",
    experiment_assignments_df[
        [
            "experiment_id",
            "user_id"
        ]
    ].duplicated().sum()
)
```
