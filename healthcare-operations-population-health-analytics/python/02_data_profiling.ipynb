```python
completeness_records = []

for table_name, df in datasets.items():

    total_rows = len(df)

    for column in df.columns:

        missing_count = df[column].isna().sum()

        missing_pct = (
            missing_count / total_rows * 100
            if total_rows > 0
            else 0
        )

        completeness_records.append({
            "Table": table_name,
            "Column": column,
            "TotalRows": total_rows,
            "MissingCount": missing_count,
            "MissingPercent": round(missing_pct, 2)
        })

completeness_df = pd.DataFrame(
    completeness_records
)

completeness_df.head(20)
```

```python
expected_date_columns = {
    "patients": [
        "BirthDate",
        "RegistrationDate"
    ],
    "providers": [
        "HireDate"
    ],
    "encounters": [
        "EncounterDate",
        "ArrivalDateTime",
        "TriageDateTime",
        "ProviderStartDateTime",
        "EncounterEndDateTime"
    ],
    "encounter_procedures": [
        "ProcedureDate"
    ],
    "admissions": [
        "AdmissionDateTime",
        "DischargeDateTime",
        "FollowUpDate"
    ],
    "appointments": [
        "ScheduledDate",
        "AppointmentDateTime"
    ],
    "lab_results": [
        "ResultDateTime"
    ]
}
```

```python
date_type_profile = []

for table_name, columns in expected_date_columns.items():
    df = datasets[table_name]

    for column in columns:
        date_type_profile.append({
            "Table": table_name,
            "Column": column,
            "CurrentDataType": str(df[column].dtype),
            "ExpectedDataType": "datetime"
        })

date_type_profile_df = pd.DataFrame(
    date_type_profile
)

date_type_profile_df
```

```python
date_parse_results = []

for table_name, columns in expected_date_columns.items():

    df = datasets[table_name]

    for column in columns:

        converted = pd.to_datetime(
            df[column],
            errors="coerce"
        )

        original_non_null = df[column].notna().sum()
        converted_non_null = converted.notna().sum()

        failed_parse_count = (
            original_non_null
            - converted_non_null
        )

        date_parse_results.append({
            "Table": table_name,
            "Column": column,
            "OriginalNonNull": original_non_null,
            "FailedParseCount": failed_parse_count
        })

date_parse_profile_df = pd.DataFrame(
    date_parse_results
)

date_parse_profile_df
```

```python
date_parse_issues_df = date_parse_profile_df[
    date_parse_profile_df["FailedParseCount"] > 0
]

date_parse_issues_df
```

```python
numeric_columns = {
    "procedures": [
        "StandardCost"
    ],
    "encounters": [
        "EncounterCost"
    ],
    "encounter_procedures": [
        "ProcedureCost"
    ],
    "lab_results": [
        "ResultValue",
        "ReferenceLow",
        "ReferenceHigh"
    ]
}
```

```python
numeric_type_profile = []

for table_name, columns in numeric_columns.items():

    df = datasets[table_name]

    for column in columns:

        numeric_type_profile.append({
            "Table": table_name,
            "Column": column,
            "CurrentDataType": str(df[column].dtype),
            "ExpectedType": "numeric"
        })

numeric_type_profile_df = pd.DataFrame(
    numeric_type_profile
)

numeric_type_profile_df
```

```python
numeric_parse_results = []

for table_name, columns in numeric_columns.items():

    df = datasets[table_name]

    for column in columns:

        converted = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        original_non_null = df[column].notna().sum()
        converted_non_null = converted.notna().sum()

        failed_numeric_parse = (
            original_non_null
            - converted_non_null
        )

        numeric_parse_results.append({
            "Table": table_name,
            "Column": column,
            "FailedNumericParseCount": failed_numeric_parse
        })

numeric_parse_profile_df = pd.DataFrame(
    numeric_parse_results
)

numeric_parse_profile_df
```

```python
relationship_checks = [
    ("providers", "DepartmentID", "departments", "DepartmentID"),

    ("encounters", "PatientID", "patients", "PatientID"),
    ("encounters", "ProviderID", "providers", "ProviderID"),
    ("encounters", "DepartmentID", "departments", "DepartmentID"),
    ("encounters", "PayerID", "payers", "PayerID"),

    ("encounter_diagnoses", "EncounterID", "encounters", "EncounterID"),
    ("encounter_diagnoses", "DiagnosisID", "diagnoses", "DiagnosisID"),

    ("encounter_procedures", "EncounterID", "encounters", "EncounterID"),
    ("encounter_procedures", "ProcedureID", "procedures", "ProcedureID"),

    ("admissions", "EncounterID", "encounters", "EncounterID"),
    ("admissions", "PatientID", "patients", "PatientID"),
    ("admissions", "DepartmentID", "departments", "DepartmentID"),

    ("appointments", "PatientID", "patients", "PatientID"),
    ("appointments", "ProviderID", "providers", "ProviderID"),
    ("appointments", "DepartmentID", "departments", "DepartmentID"),

    ("lab_results", "EncounterID", "encounters", "EncounterID"),
    ("lab_results", "PatientID", "patients", "PatientID")
]
```

```python
ri_results = []

for child_table, child_fk, parent_table, parent_pk in relationship_checks:

    child_df = datasets[child_table]
    parent_df = datasets[parent_table]

    non_null_child = child_df[
        child_df[child_fk].notna()
    ]

    invalid_mask = (
        ~non_null_child[child_fk]
        .isin(parent_df[parent_pk])
    )

    invalid_count = invalid_mask.sum()

    ri_results.append({
        "ChildTable": child_table,
        "ForeignKey": child_fk,
        "ParentTable": parent_table,
        "ParentKey": parent_pk,
        "CheckedRows": len(non_null_child),
        "InvalidReferenceCount": invalid_count,
        "InvalidReferencePercent": round(
            invalid_count / len(non_null_child) * 100,
            3
        ) if len(non_null_child) > 0 else 0
    })

ri_profile_df = pd.DataFrame(ri_results)

ri_profile_df
```

```python
ri_issues_df = ri_profile_df[
    ri_profile_df[
        "InvalidReferenceCount"
    ] > 0
].sort_values(
    "InvalidReferenceCount",
    ascending=False
)

ri_issues_df
```

```python
lab_outlier_records = []

for lab_test, group in lab_results.groupby(
    "LabTest"
):

    q1 = group[
        "ResultValue"
    ].quantile(0.25)

    q3 = group[
        "ResultValue"
    ].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 3 * iqr
    upper_bound = q3 + 3 * iqr

    outlier_count = (
        (
            group["ResultValue"]
            < lower_bound
        )
        |
        (
            group["ResultValue"]
            > upper_bound
        )
    ).sum()

    lab_outlier_records.append({
        "LabTest": lab_test,
        "LowerBound": round(
            lower_bound,
            2
        ),
        "UpperBound": round(
            upper_bound,
            2
        ),
        "ExtremeOutlierCount": int(
            outlier_count
        )
    })

lab_outlier_profile_df = pd.DataFrame(
    lab_outlier_records
)

lab_outlier_profile_df
```
