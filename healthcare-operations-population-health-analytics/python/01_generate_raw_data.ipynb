```python
department_ids = departments_df["DepartmentID"].tolist()

provider_department_assignments = []

# Give every department at least 4 providers
for department_id in department_ids:
    provider_department_assignments.extend([department_id] * 4)

print("Providers after minimum assignment:",
      len(provider_department_assignments))
```

```python
additional_departments = np.random.choice(
    department_ids,
    size=20,
    replace=True
)

provider_department_assignments.extend(additional_departments)

print("Total provider assignments:",
      len(provider_department_assignments))
```

```python
provider_records = []

hire_start = datetime(2012, 1, 1)
hire_end = datetime(2025, 12, 31)

hire_range_days = (hire_end - hire_start).days

for i in range(120):

    provider_id = f"PRV{i + 1:03d}"

    department_id = provider_department_assignments[i]

    department_name = departments_df.loc[
        departments_df["DepartmentID"] == department_id,
        "DepartmentName"
    ].iloc[0]

    specialty = specialty_map[department_name]

    provider_type = np.random.choice(
        provider_types,
        p=provider_type_probabilities
    )

    random_days = random.randint(0, hire_range_days)

    hire_date = (
        hire_start + timedelta(days=random_days)
    ).date()

    active_flag = np.random.choice(
        [1, 0],
        p=[0.92, 0.08]
    )

    provider_records.append(
        (
            provider_id,
            department_id,
            provider_type,
            specialty,
            hire_date,
            active_flag
        )
    )
```

```python
patient_count = 10_000

gender_values = [
    "Male",
    "Female",
    "Other",
    "Unknown"
]

gender_probabilities = [
    0.485,
    0.495,
    0.01,
    0.01
]

race_values = [
    "White",
    "Black or African American",
    "Asian",
    "American Indian or Alaska Native",
    "Native Hawaiian or Other Pacific Islander",
    "Multiple",
    "Other",
    "Unknown"
]

race_probabilities = [
    0.50,
    0.25,
    0.10,
    0.02,
    0.01,
    0.05,
    0.04,
    0.03
]

ethnicity_values = [
    "Hispanic or Latino",
    "Not Hispanic or Latino",
    "Unknown"
]

ethnicity_probabilities = [
    0.12,
    0.85,
    0.03
]
```

```python
region_zip_map = {
    "Central": ["23220", "23221", "23223", "23225"],
    "Northern": ["22401", "22405", "22554", "22551"],
    "Southern": ["23803", "23805", "23831", "23834"],
    "Eastern": ["23185", "23601", "23602", "23606"],
    "Western": ["22901", "22902", "22903", "22911"]
}

region_values = [
    "Central",
    "Northern",
    "Southern",
    "Eastern",
    "Western"
]

region_probabilities = [
    0.35,
    0.20,
    0.18,
    0.12,
    0.15
]
```

```python
age_groups = [
    (0, 17),
    (18, 34),
    (35, 49),
    (50, 64),
    (65, 90)
]

age_group_probabilities = [
    0.18,
    0.20,
    0.20,
    0.22,
    0.20
]
```

```python
patient_records = []

reference_date = datetime(2025, 12, 31).date()

registration_start = datetime(2023, 1, 1).date()
registration_end = datetime(2025, 12, 31).date()

for i in range(patient_count):

    patient_id = f"PAT{i + 1:06d}"

    # Select age group
    age_group_index = np.random.choice(
        len(age_groups),
        p=age_group_probabilities
    )

    min_age, max_age = age_groups[age_group_index]

    age = random.randint(min_age, max_age)

    # Approximate DOB and randomize within the year
    birth_year = reference_date.year - age

    birth_date = datetime(
        birth_year,
        random.randint(1, 12),
        1
    ).date()

    # Safely choose a valid day within the month
    birth_date = birth_date + timedelta(
        days=random.randint(0, 27)
    )

    gender = np.random.choice(
        gender_values,
        p=gender_probabilities
    )

    race = np.random.choice(
        race_values,
        p=race_probabilities
    )

    ethnicity = np.random.choice(
        ethnicity_values,
        p=ethnicity_probabilities
    )

    region = np.random.choice(
        region_values,
        p=region_probabilities
    )

    zip_code = random.choice(
        region_zip_map[region]
    )

    state = "VA"

    # Registration cannot occur before birth
    earliest_registration = max(
        registration_start,
        birth_date
    )

    registration_range = (
        registration_end - earliest_registration
    ).days

    registration_date = (
        earliest_registration
        + timedelta(
            days=random.randint(0, registration_range)
        )
    )

    patient_records.append(
        (
            patient_id,
            birth_date,
            gender,
            race,
            ethnicity,
            zip_code,
            state,
            region,
            registration_date
        )
    )
```

```python
def choose_department(age):

    weights = department_generation_df["BaseWeight"].astype(float).copy()

    # Children use Pediatrics more frequently
    if age < 18:
        pediatric_mask = (
            department_generation_df["DepartmentName"]
            == "Pediatrics"
        )

        weights.loc[pediatric_mask] *= 5.0

        cardiology_mask = (
            department_generation_df["DepartmentName"]
            == "Cardiology"
        )

        weights.loc[cardiology_mask] *= 0.40

    # Older adults use more inpatient, ED, and cardiology services
    elif age >= 65:

        general_medicine_mask = (
            department_generation_df["DepartmentName"]
            == "General Medicine"
        )

        cardiology_mask = (
            department_generation_df["DepartmentName"]
            == "Cardiology"
        )

        emergency_mask = (
            department_generation_df["DepartmentName"]
            == "Emergency Department"
        )

        weights.loc[general_medicine_mask] *= 1.70
        weights.loc[cardiology_mask] *= 1.50
        weights.loc[emergency_mask] *= 1.30

    probabilities = weights / weights.sum()

    return np.random.choice(
        department_generation_df["DepartmentID"],
        p=probabilities
    )
```

```python
def choose_encounter_type(department_name, department_type):

    if department_name == "Emergency Department":
        return np.random.choice(
            ["Emergency", "Urgent Care"],
            p=[0.90, 0.10]
        )

    if department_type == "Inpatient":
        return "Inpatient"

    if department_type == "Primary Care":
        return "Primary Care"

    if department_type == "Outpatient":
        return np.random.choice(
            ["Outpatient", "Specialist"],
            p=[0.55, 0.45]
        )

    if department_type in [
        "Specialty",
        "Behavioral Health"
    ]:
        return "Specialist"

    return "Outpatient"
```

```python
def choose_payer(age):

    payer_names = [
        "Medicare",
        "Medicaid",
        "Commercial Insurance",
        "Employer Sponsored",
        "Self Pay",
        "Other"
    ]

    if age >= 65:

        probabilities = [
            0.65,
            0.10,
            0.10,
            0.05,
            0.05,
            0.05
        ]

    elif age < 18:

        probabilities = [
            0.00,
            0.30,
            0.25,
            0.30,
            0.10,
            0.05
        ]

    else:

        probabilities = [
            0.05,
            0.15,
            0.25,
            0.35,
            0.15,
            0.05
        ]

    payer_name = np.random.choice(
        payer_names,
        p=probabilities
    )

    return payer_name_to_id[payer_name]
```

```python
def generate_arrival_datetime(encounter_date, encounter_type):

    if encounter_type in ["Emergency", "Urgent Care"]:
        hour = random.randint(0, 23)

    elif encounter_type == "Inpatient":
        hour = random.randint(6, 22)

    else:
        hour = random.randint(7, 17)

    minute = random.randint(0, 59)

    return datetime.combine(
        encounter_date,
        datetime.min.time()
    ).replace(
        hour=hour,
        minute=minute
    )
```

```python
def generate_wait_minutes(
    encounter_type,
    facility_name,
    department_name,
    arrival_datetime
):

    base_wait = {
        "Emergency": 60,
        "Urgent Care": 35,
        "Inpatient": 45,
        "Primary Care": 15,
        "Outpatient": 20,
        "Specialist": 25
    }

    target = base_wait[encounter_type]

    wait_minutes = int(
        np.random.gamma(
            shape=2.0,
            scale=target / 2
        )
    )

    # Simulate an operational bottleneck
    # at RiverCare Central Hospital ED.
    if (
        facility_name == "RiverCare Central Hospital"
        and department_name == "Emergency Department"
    ):
        wait_minutes += 25

    # Evening ED demand
    if (
        encounter_type == "Emergency"
        and arrival_datetime.hour >= 17
    ):
        wait_minutes += 15

    # Weekend pressure
    if arrival_datetime.weekday() >= 5:
        wait_minutes += 10

    return max(wait_minutes, 3)
```

```python
def generate_encounter_cost(
    encounter_type,
    duration_hours
):

    if encounter_type == "Primary Care":
        cost = np.random.normal(220, 60)

    elif encounter_type == "Outpatient":
        cost = np.random.normal(550, 180)

    elif encounter_type == "Specialist":
        cost = np.random.normal(750, 250)

    elif encounter_type == "Urgent Care":
        cost = np.random.normal(450, 150)

    elif encounter_type == "Emergency":
        cost = np.random.normal(1800, 650)

    elif encounter_type == "Inpatient":
        estimated_days = max(
            duration_hours / 24,
            1
        )

        cost = (
            4000
            + estimated_days * 2200
            + np.random.normal(0, 1200)
        )

    else:
        cost = 500

    return round(
        max(cost, 50),
        2
    )
```

```python
def choose_primary_diagnosis(
    patient_age,
    department_name,
    encounter_type
):

    # Pediatrics / younger patients
    if patient_age < 18:
        pool = (
            diagnosis_groups["respiratory"] * 4
            + diagnosis_groups["general"] * 2
            + diagnosis_groups["injury"] * 2
            + diagnosis_groups["gastrointestinal"]
        )

    # Cardiology
    elif department_name == "Cardiology":
        pool = (
            diagnosis_groups["heart"] * 6
            + diagnosis_groups["hypertension"] * 4
            + diagnosis_groups["diabetes"] * 2
            + diagnosis_groups["general"]
        )

    # Endocrinology
    elif department_name == "Endocrinology":
        pool = (
            diagnosis_groups["diabetes"] * 6
            + diagnosis_groups["general"] * 3
            + diagnosis_groups["hypertension"] * 2
        )

    # Pulmonology
    elif department_name == "Pulmonology":
        pool = (
            diagnosis_groups["respiratory"] * 6
            + diagnosis_groups["general"]
        )

    # Neurology
    elif department_name == "Neurology":
        pool = (
            diagnosis_groups["neurological"] * 6
            + diagnosis_groups["general"]
        )

    # Orthopedics
    elif department_name == "Orthopedics":
        pool = (
            diagnosis_groups["musculoskeletal"] * 5
            + diagnosis_groups["injury"] * 4
            + diagnosis_groups["general"]
        )

    # Oncology
    elif department_name == "Oncology":
        pool = (
            diagnosis_groups["cancer"] * 7
            + diagnosis_groups["general"]
        )

    # Behavioral Health
    elif department_name == "Behavioral Health":
        pool = (
            diagnosis_groups["mental_health"] * 8
            + diagnosis_groups["general"]
        )

    # Gastroenterology
    elif department_name == "Gastroenterology":
        pool = (
            diagnosis_groups["gastrointestinal"] * 7
            + diagnosis_groups["general"]
        )

    # Emergency
    elif encounter_type in ["Emergency", "Urgent Care"]:
        pool = (
            diagnosis_groups["heart"] * 2
            + diagnosis_groups["respiratory"] * 3
            + diagnosis_groups["injury"] * 3
            + diagnosis_groups["gastrointestinal"] * 2
            + diagnosis_groups["neurological"] * 2
            + diagnosis_groups["general"] * 3
        )

    # Older adults
    elif patient_age >= 65:
        pool = (
            diagnosis_groups["heart"] * 3
            + diagnosis_groups["hypertension"] * 3
            + diagnosis_groups["diabetes"] * 2
            + diagnosis_groups["kidney"] * 2
            + diagnosis_groups["respiratory"] * 2
            + diagnosis_groups["general"] * 2
        )

    # General adult population
    else:
        pool = (
            diagnosis_groups["general"] * 3
            + diagnosis_groups["hypertension"] * 2
            + diagnosis_groups["diabetes"]
            + diagnosis_groups["respiratory"] * 2
            + diagnosis_groups["musculoskeletal"] * 2
            + diagnosis_groups["mental_health"]
            + diagnosis_groups["gastrointestinal"]
        )

    return random.choice(pool)
```

```python
def choose_secondary_diagnosis(
    patient_age,
    primary_diagnosis
):

    if patient_age >= 65:
        pool = (
            diagnosis_groups["hypertension"] * 4
            + diagnosis_groups["diabetes"] * 3
            + diagnosis_groups["heart"] * 3
            + diagnosis_groups["kidney"] * 2
            + diagnosis_groups["respiratory"] * 2
            + diagnosis_groups["general"]
        )

    elif patient_age >= 50:
        pool = (
            diagnosis_groups["hypertension"] * 3
            + diagnosis_groups["diabetes"] * 2
            + diagnosis_groups["heart"] * 2
            + diagnosis_groups["general"] * 2
            + diagnosis_groups["musculoskeletal"]
        )

    elif patient_age >= 18:
        pool = (
            diagnosis_groups["general"] * 3
            + diagnosis_groups["mental_health"]
            + diagnosis_groups["respiratory"]
            + diagnosis_groups["musculoskeletal"]
            + diagnosis_groups["hypertension"]
        )

    else:
        pool = (
            diagnosis_groups["respiratory"] * 3
            + diagnosis_groups["general"] * 2
        )

    # Avoid immediately returning the primary diagnosis
    filtered_pool = [
        diagnosis_id
        for diagnosis_id in pool
        if diagnosis_id != primary_diagnosis
    ]

    return random.choice(filtered_pool)
```

```python
procedure_counts = [0, 1, 2]

procedure_count_probabilities = [
    0.35,
    0.52,
    0.13
]
```

```python
def generate_actual_procedure_cost(procedure_id):

    standard_cost = procedure_standard_cost[
        procedure_id
    ]

    multiplier = np.random.normal(
        loc=1.00,
        scale=0.08
    )

    actual_cost = (
        standard_cost * multiplier
    )

    return round(
        max(actual_cost, 10),
        2
    )
```

```python
encounter_procedure_records = []

procedure_counter = 1

for _, encounter in encounters_clean_df.iterrows():

    encounter_id = encounter["EncounterID"]

    department_id = encounter[
        "DepartmentID"
    ]

    encounter_type = encounter[
        "EncounterType"
    ]

    department_name = (
        department_info[
            department_id
        ]["DepartmentName"]
    )

    procedure_count = np.random.choice(
        procedure_counts,
        p=procedure_count_probabilities
    )

    if procedure_count == 0:
        continue

    procedure_pool = get_procedure_pool(
        department_name,
        encounter_type
    )

    selected_procedures = set()

    for _ in range(procedure_count):

        procedure_id = random.choice(
            procedure_pool
        )

        attempts = 0

        while (
            procedure_id in selected_procedures
            and attempts < 10
        ):
            procedure_id = random.choice(
                procedure_pool
            )

            attempts += 1

        if procedure_id in selected_procedures:
            continue

        selected_procedures.add(
            procedure_id
        )

        arrival_datetime = pd.Timestamp(
            encounter["ArrivalDateTime"]
        )

        encounter_end_datetime = pd.Timestamp(
            encounter["EncounterEndDateTime"]
        )

        total_minutes = max(
            int(
                (
                    encounter_end_datetime
                    - arrival_datetime
                ).total_seconds()
                / 60
            ),
            1
        )

        procedure_offset_minutes = random.randint(
            0,
            total_minutes
        )

        procedure_datetime = (
            arrival_datetime
            + timedelta(
                minutes=procedure_offset_minutes
            )
        )

        procedure_date = (
            procedure_datetime.date()
        )

        procedure_cost = (
            generate_actual_procedure_cost(
                procedure_id
            )
        )

        encounter_procedure_records.append(
            (
                f"EPR{procedure_counter:07d}",
                encounter_id,
                procedure_id,
                procedure_date,
                procedure_cost
            )
        )

        procedure_counter += 1
```

```python
admission_records = []

for i, (_, encounter) in enumerate(
    inpatient_encounters_df.iterrows(),
    start=1
):

    admission_id = f"ADM{i:06d}"

    encounter_id = encounter["EncounterID"]
    patient_id = encounter["PatientID"]
    department_id = encounter["DepartmentID"]

    patient_age = patient_age_lookup[
        patient_id
    ]

    admission_datetime = pd.Timestamp(
        encounter["ArrivalDateTime"]
    )

    discharge_datetime = pd.Timestamp(
        encounter["EncounterEndDateTime"]
    )

    admission_type = np.random.choice(
        admission_types,
        p=admission_type_probabilities
    )

    discharge_disposition = (
        choose_discharge_disposition(
            patient_age
        )
    )

    chronic_flag = (
        primary_chronic_lookup.get(
            encounter_id,
            0
        )
    )

    # ---------------------------------
    # Follow-up requirement
    # ---------------------------------

    if (
        chronic_flag == 1
        or patient_age >= 65
    ):
        follow_up_required_probability = 0.85
    else:
        follow_up_required_probability = 0.65

    follow_up_required = (
        1
        if random.random()
        < follow_up_required_probability
        else 0
    )

    # ---------------------------------
    # Follow-up completion
    # ---------------------------------

    if follow_up_required == 1:

        discharge_year = (
            discharge_datetime.year
        )

        # Slightly declining follow-up
        # performance over time
        if discharge_year == 2023:
            completion_probability = 0.78

        elif discharge_year == 2024:
            completion_probability = 0.72

        else:
            completion_probability = 0.66

        follow_up_completed = (
            1
            if random.random()
            < completion_probability
            else 0
        )

    else:
        follow_up_completed = 0

    # ---------------------------------
    # Follow-up date
    # ---------------------------------

    if follow_up_completed == 1:

        days_to_follow_up = int(
            np.random.choice(
                [
                    3,
                    5,
                    7,
                    10,
                    14,
                    21,
                    30
                ],
                p=[
                    0.08,
                    0.12,
                    0.25,
                    0.18,
                    0.20,
                    0.10,
                    0.07
                ]
            )
        )

        follow_up_date = (
            discharge_datetime
            + timedelta(
                days=days_to_follow_up
            )
        ).date()

    else:
        follow_up_date = None

    admission_records.append(
        (
            admission_id,
            encounter_id,
            patient_id,
            department_id,
            admission_datetime,
            discharge_datetime,
            admission_type,
            discharge_disposition,
            follow_up_required,
            follow_up_completed,
            follow_up_date
        )
    )
```

```python
def generate_booking_lead_days(appointment_type):

    if appointment_type == "Post-Discharge Follow-Up":
        return random.randint(1, 14)

    elif appointment_type == "Follow-Up":
        return random.randint(3, 45)

    elif appointment_type == "New Patient":
        return random.randint(5, 60)

    elif appointment_type == "Specialist":
        return random.randint(7, 75)

    else:
        return random.randint(1, 40)
```

```python
def choose_appointment_status(
    booking_lead_days,
    appointment_type
):

    no_show_prob = 0.08
    cancelled_prob = 0.10
    rescheduled_prob = 0.06

    if booking_lead_days > 30:
        no_show_prob += 0.05

    if booking_lead_days > 60:
        no_show_prob += 0.04

    if appointment_type == "Post-Discharge Follow-Up":
        no_show_prob -= 0.02

    no_show_prob = max(no_show_prob, 0.02)

    completed_prob = (
        1
        - no_show_prob
        - cancelled_prob
        - rescheduled_prob
    )

    return np.random.choice(
        [
            "Completed",
            "Cancelled",
            "No Show",
            "Rescheduled"
        ],
        p=[
            completed_prob,
            cancelled_prob,
            no_show_prob,
            rescheduled_prob
        ]
    )
```

```python
cancellation_reasons = [
    "Patient Request",
    "Provider Unavailable",
    "Scheduling Conflict",
    "Insurance Issue",
    "Transportation Issue",
    "Other"
]
```

```python
appointment_records = []

for i in range(appointment_count):

    appointment_id = f"APT{i + 1:06d}"

    # -------------------------
    # Patient
    # -------------------------

    patient_index = np.random.choice(
        patients_generation_df.index,
        p=patient_probabilities
    )

    patient_row = patients_generation_df.loc[
        patient_index
    ]

    patient_id = patient_row["PatientID"]

    registration_date = (
        patient_row["RegistrationDate"].date()
    )

    # -------------------------
    # Department
    # -------------------------

    department_id = random.choice(
        appointment_department_ids
    )

    # -------------------------
    # Provider
    # -------------------------

    provider_id = random.choice(
        providers_by_department[
            department_id
        ]
    )

    # -------------------------
    # Appointment Type
    # -------------------------

    appointment_type = np.random.choice(
        appointment_types,
        p=appointment_type_probabilities
    )

    # -------------------------
    # Scheduled Date
    # -------------------------

    latest_schedule_date = datetime(
        2025,
        11,
        30
    ).date()

    days_available = (
        latest_schedule_date
        - registration_date
    ).days

    scheduled_date = (
        registration_date
        + timedelta(
            days=random.randint(
                0,
                max(days_available, 0)
            )
        )
    )

    # -------------------------
    # Booking Lead Time
    # -------------------------

    booking_lead_days = (
        generate_booking_lead_days(
            appointment_type
        )
    )

    appointment_date = (
        scheduled_date
        + timedelta(
            days=booking_lead_days
        )
    )

    # Keep inside study period
    if appointment_date > datetime(
        2025,
        12,
        31
    ).date():

        appointment_date = datetime(
            2025,
            12,
            31
        ).date()

    # -------------------------
    # Appointment Time
    # -------------------------

    appointment_hour = random.randint(
        8,
        16
    )

    appointment_minute = random.choice(
        [0, 15, 30, 45]
    )

    appointment_datetime = datetime.combine(
        appointment_date,
        datetime.min.time()
    ).replace(
        hour=appointment_hour,
        minute=appointment_minute
    )

    # -------------------------
    # Status
    # -------------------------

    actual_lead_days = (
        appointment_date
        - scheduled_date
    ).days

    appointment_status = (
        choose_appointment_status(
            actual_lead_days,
            appointment_type
        )
    )

    # -------------------------
    # Cancellation Reason
    # -------------------------

    if appointment_status == "Cancelled":
        cancellation_reason = random.choice(
            cancellation_reasons
        )
    else:
        cancellation_reason = None

    appointment_records.append(
        (
            appointment_id,
            patient_id,
            provider_id,
            department_id,
            scheduled_date,
            appointment_datetime,
            appointment_status,
            appointment_type,
            cancellation_reason
        )
    )
```

```python
lab_test_config = {
    "HbA1c": {
        "unit": "%",
        "low": 4.0,
        "high": 5.6,
        "mean": 6.2,
        "std": 1.3
    },

    "Glucose": {
        "unit": "mg/dL",
        "low": 70,
        "high": 99,
        "mean": 105,
        "std": 30
    },

    "LDL Cholesterol": {
        "unit": "mg/dL",
        "low": 0,
        "high": 100,
        "mean": 115,
        "std": 35
    },

    "Creatinine": {
        "unit": "mg/dL",
        "low": 0.6,
        "high": 1.3,
        "mean": 1.0,
        "std": 0.3
    },

    "Hemoglobin": {
        "unit": "g/dL",
        "low": 12.0,
        "high": 17.5,
        "mean": 14.2,
        "std": 1.8
    }
}
```

```python
def classify_lab_result(value, low, high):

    if value < low:
        return "Low"

    elif value > high * 1.5:
        return "Critical"

    elif value > high:
        return "High"

    else:
        return "Normal"
```

```python
lab_result_records = []

lab_counter = 1

eligible_encounters = encounters_clean_df[
    encounters_clean_df["EncounterStatus"] != "Left Without Being Seen"
].copy()

while len(lab_result_records) < 100_000:

    encounter = eligible_encounters.sample(1).iloc[0]

    encounter_id = encounter["EncounterID"]
    patient_id = encounter["PatientID"]

    lab_test = random.choice(
        list(lab_test_config.keys())
    )

    config = lab_test_config[lab_test]

    result_value = np.random.normal(
        config["mean"],
        config["std"]
    )

    result_value = round(
        max(result_value, 0.01),
        2
    )

    reference_low = config["low"]
    reference_high = config["high"]

    result_flag = classify_lab_result(
        result_value,
        reference_low,
        reference_high
    )

    arrival = pd.Timestamp(
        encounter["ArrivalDateTime"]
    )

    encounter_end = pd.Timestamp(
        encounter["EncounterEndDateTime"]
    )

    total_minutes = max(
        int(
            (encounter_end - arrival)
            .total_seconds() / 60
        ),
        1
    )

    result_datetime = (
        arrival
        + timedelta(
            minutes=random.randint(
                0,
                total_minutes
            )
        )
    )

    lab_result_records.append(
        (
            f"LAB{lab_counter:07d}",
            encounter_id,
            patient_id,
            lab_test,
            result_value,
            config["unit"],
            reference_low,
            reference_high,
            result_flag,
            result_datetime
        )
    )

    lab_counter += 1
```

```python
lab_results_clean_df = pd.DataFrame(
    lab_result_records,
    columns=[
        "LabResultID",
        "EncounterID",
        "PatientID",
        "LabTest",
        "ResultValue",
        "ResultUnit",
        "ReferenceLow",
        "ReferenceHigh",
        "ResultFlag",
        "ResultDateTime"
    ]
)

lab_results_clean_df.head()
```

```python
invalid_unit_count = int(
    len(lab_results_df) * 0.003
)

available_indices = lab_results_df.index.difference(
    missing_unit_indices
)

invalid_unit_indices = np.random.choice(
    available_indices,
    size=invalid_unit_count,
    replace=False
)

lab_results_df.loc[
    invalid_unit_indices,
    "ResultUnit"
] = "INVALID_UNIT"
```

```python
outlier_count = int(
    len(lab_results_df) * 0.003
)

outlier_indices = np.random.choice(
    lab_results_df.index,
    size=outlier_count,
    replace=False
)

lab_results_df.loc[
    outlier_indices,
    "ResultValue"
] = (
    lab_results_df.loc[
        outlier_indices,
        "ResultValue"
    ] * 50
)
```
